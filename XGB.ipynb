```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

## 2. Data Collection

```python
breast_data.target
```

```python
breast_data = load_breast_cancer()
breast_data_df = pd.DataFrame(breast_data.data,columns=breast_data.feature_names)
breast_data_df["Target"] = breast_data.target
```

## 4. Data Preparation

```python
x = breast_data_df.drop(labels="Target",axis=1)
x
```

```python
y = breast_data_df["Target"]
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.20,random_state=123,stratify=y)
```

## 5. Model Building

```python
xgb_model = xgb.XGBClassifier(n_estimators=30,max_depth=3,learning_rate=0.02,gamma=1.2)
```

## 6. Model Training

```python
xgb_model.fit(X_train,y_train)
```

## 7. Model Testing

```python
y_tr_pred = xgb_model.predict(X_train)
```

```python
y_te_pred = xgb_model.predict(X_test)
```

## 8. Model Evaluation

```python
print(classification_report(y_train,y_tr_pred))
```

```python
confusion_matrix(y_train,y_tr_pred)
```

```python
print(classification_report(y_test,y_te_pred))
```

```python
confusion_matrix(y_test,y_te_pred)
```
